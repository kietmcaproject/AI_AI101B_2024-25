```python
# Import libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import confusion_matrix, accuracy_score, classification_report, roc_auc_score, ConfusionMatrixDisplay
from sklearn.ensemble import IsolationForest, RandomForestClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.svm import SVC

# Loading the dataset to a Pandas Dataframe
credit_card_data = pd.read_csv('/content/creditcard.csv')

# Let's see first 5 rows of the dataset
print(credit_card_data.head(5))

# Let's see last 5 rows of our dataset
print(credit_card_data.tail())

# Dataset information
print(credit_card_data.info())

# Checking number of missing values
print(credit_card_data.isnull().sum())

# Find distribution of Normal transaction or Fraud transaction
print(credit_card_data['Class'].value_counts())

# Separating the data
normal = credit_card_data[credit_card_data.Class == 0]
fraud = credit_card_data[credit_card_data.Class == 1]

print(normal.shape)
print(fraud.shape)

# Visualize the data
labels = ["Normal", "Fraud"]
count_classes = credit_card_data['Class'].value_counts()
count_classes.plot(kind="bar", rot=0)
plt.title("ProjectGurukul")

plt.ylabel("Count")
plt.xticks(range(2), labels)
plt.show()

# Statistical measures of the data
print(normal.Amount.describe())
print(fraud.Amount.describe())

# Visualize the data
sns.relplot(x='Amount', y='Time',	hue='Class', data=credit_card_data)

# Compare values of both transactions
print(credit_card_data.groupby('Class').mean())

# Now we will build a sample dataset containing similar distribution of normal transaction and fraud transaction
normal_sample = normal.sample(n=492)

# Concatenate two dataframes
credit_card_new_data	=	pd.concat([normal_sample,	fraud], axis=0)
print(credit_card_new_data['Class'].value_counts()) # Splitting data into features and targets
X = credit_card_new_data.drop('Class', axis=1)
Y = credit_card_new_data['Class']

# Splitting the data into training and testing data
X_train,	X_test,	Y_train,	Y_test	=	train_test_split(X,	Y, test_size=0.2, stratify=Y, random_state=2)
print(X.shape, X_train.shape, X_test.shape) # Creating Model
model = LogisticRegression()

# Training the Logistic Regression model with training data model.fit(X_train, Y_train)
# === Models ===
models = {
"Logistic Regression": LogisticRegression(),
"Decision Tree": DecisionTreeClassifier(random_state=2),
"Random Forest": RandomForestClassifier(random_state=2),
"Support	Vector	Machine":	SVC(probability=True, random_state=2)
}

# Train and evaluate each model
for name, model in models.items():
  print(f"\n=== {name} ===")
  model.fit(X_train, Y_train)

  # Predict
  Y_train_pred = model.predict(X_train)
  Y_test_pred = model.predict(X_test)
  Y_test_proba = model.predict_proba(X_test)[:, 1]

  # Metrics
  train_acc = accuracy_score(Y_train, Y_train_pred)
  test_acc = accuracy_score(Y_test, Y_test_pred)
  roc_auc = roc_auc_score(Y_test, Y_test_proba)

  print("Training Accuracy:", train_acc)
  print("Testing Accuracy :", test_acc)
  print("ROC-AUC Score	:", roc_auc)

  # Confusion Matrix
  cm = confusion_matrix(Y_test, Y_test_pred)
  disp	=	ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=['Normal', 'Fraud'])
  disp.plot()
  plt.title(f"Confusion Matrix - {name}")
  plt.show()

  # Model Evaluation
  X_train_pred = model.predict(X_train)
  training_data_accuracy = accuracy_score(X_train_pred, Y_train)
  print('Accuracy of Training data:', training_data_accuracy)

  # Accuracy on test data
  X_test_pred = model.predict(X_test)
  test_data_accuracy = accuracy_score(X_test_pred, Y_test)
  print('Accuracy of Testing data:', test_data_accuracy)

  # Confusion matrix and classification report
  print(confusion_matrix(Y_test, X_test_pred))
  print(classification_report(Y_test, X_test_pred))

# Anomaly Detection using Isolation Forest
iso_forest	=	IsolationForest(contamination=0.01, random_state=42)
X_iso = credit_card_data.drop('Class', axis=1)	# Use features only
y_iso = iso_forest.fit_predict(X_iso)

# Convert predictions to binary (1 for normal, -1 for anomaly)
y_iso = np.where(y_iso == -1, 1, 0)

# Add predictions to the original dataframe
credit_card_data['Anomaly'] = y_iso

# Visualize the anomalies
plt.figure(figsize=(10, 6))
sns.scatterplot(data=credit_card_data,	x='Amount',	y='Time', hue='Anomaly', palette={0: 'blue', 1: 'red'}, alpha=0.5)
plt.title('Anomaly Detection in Credit Card Transactions')
plt.xlabel('Amount')
plt.ylabel('Time')
plt.legend(title='Anomaly',	loc='upper right',	labels=['Normal', 'Anomaly'])
plt.show()

# Plot combined ROC curve
plt.plot([0, 1], [0, 1], linestyle='--', color='gray')
plt.title("ROC Curve Comparison")
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.legend()
plt.grid()
plt.show()
```
